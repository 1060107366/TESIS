## Iniciar sesion
POST::http://127.0.0.1:5001/auth/login

> Obtienes el acces_token *auth::Bearer=token*
```JSON
{
    "email": "alpha01@example.com",
    "contrasena": "123456"
}
```
#este usuario contiene los clientes de id = 8 hasta el id = 52

### *Crear cliente*
> auth::Bearer=token\
> POST==http://127.0.0.1:5001/customer/create\

Puedes decidir si poner un valor total y funcione el metodo como (creacion y compra)
o quitar el campo valor_orden_total y solo crear el usuario(no recomendado ya que el campo ultimacompra queda null)
```JSON
{
    "nombre": "Tinker wood",
    "email": "tikervelood@example.com",
    "telefono": "3130001991",
    "fecha_creacion": "2025-01-20 12:00:00",
    "valor_orden_total": 40000
}
```

### *Obtener clientes*
> auth::Bearer=token\

> GET==http://127.0.0.1:5001/customer/
