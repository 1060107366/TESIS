### Librerias

```python
import os
import pandas as pd
```

### Definir la carpeta donde se encuentran los archivos, las rutas de los archivos CSV a combinar Y Verificar si los archivos existen  y Cargar los datos en DataFrames

```python
data_dir = r"C:\Users\Sr. Sebastian\Documents\GitHub\CRMIA\app\ia\segments\Data"

# rutas de los archivos
file1 = os.path.join(data_dir, "kmeans_m1_data.csv")
file2 = os.path.join(data_dir, "kmeans_m2_data.csv")

# Verificar si los archivos existen
if not os.path.exists(file1) or not os.path.exists(file2):
    raise FileNotFoundError("Uno o ambos archivos CSV no se encuentran en el directorio especificado.")

# Cargar los datos en DataFrames
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)
```

### Verificar si las columnas son idénticas en ambos archivos Y Combinar ambos DataFrames (concatenar filas)

```python
# columnas son idénticas
if list(df1.columns) != list(df2.columns):
    raise ValueError("Las columnas de los archivos CSV no coinciden. Verifica que ambos archivos tengan la misma estructura.")

# Combinar
df_combined = pd.concat([df1, df2], ignore_index=True)
```

### Guardar el DataFrame combinado en un nuevo archivo CSV

```python
# Definir la ruta del archivo de salida
output_file = os.path.join(data_dir, "X_kmeans_data.csv")

# Guardar
df_combined.to_csv(output_file, index=False)

# Mensaje de éxito
print(f"\nArchivo combinado guardado exitosamente en: {output_file}")
print(df_combined.columns)
```
